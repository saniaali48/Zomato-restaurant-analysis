Zomato data analysis project

import libraries and create a dataframe

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
df=pd.read_csv("Zomato data .csv")
print(df)
```

```python
print(df.describe())
```

```python
print(df.info())
```

Convert Data Type of Column = rate

```python
df['rate']=df['rate'].str.split('/').str[0].astype(float)
print(df.head())
```

```python
df.info()
```

Type of Restaurant

```python
print(df.head())
```

```python
sns.countplot(x=df['listed_in(type)'])
plt.xlabel("Type of Restaurant")
```

Conclusion - majority of the resturant falls in dinning category

```python
df.head()
```

```python
df.groupby('listed_in(type)')['votes'].sum().plot(kind='line', marker='o', color='green', figsize=(6,4))
plt.xlabel('Type of Restaurant', c="red", size=20)
plt.ylabel('Votes', c="red", size=20)
plt.show()
```

#Conclusion Dinning Restaurant has recived maximum votes

```python
plt.hist(df['rate'], bins=10)
plt.title("ratings distribution")
plt.show()
```

```python
#Conclusion- the majority restaurants received ratings from 3.75 to 4.00
```

Average Orders Spending By Couples

```python
print(df.head())
```

```python
couple_data=df['approx_cost(for two people)']
sns.countplot(x=couple_data)
```

WWhich mode receives maximum rating

```python
print(df.head())
```

```python
plt.figure(figsize=(6,6))
sns.boxplot(x='online_order', y='rate', data=df)
```

Online Vs Offline Orders 

```python
pt=df.pivot_table(index='listed_in(type)', columns='online_order', aggfunc='size', fill_value=0)
plt.figure(figsize=(6,5))
sns.heatmap(pt, annot=True, cmap='YlGnBu', fmt='d')
plt.title('Heatmap')
plt.show()
```

```python
#Conclusion-Dinning restaurants primarily accept offline orders,
#whereas cafes primarily receive online orders.This suggest that clients 
#prefer orders in person at restaurants, but prefer online ordering at cafes
```
